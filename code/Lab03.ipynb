## 4.1. Debugging: a streak counter

A website wants to know the longest sequence of **consecutive days** on which visits were at least `target`. Days below the target break the streak. The function below is **deliberately incorrect**: it runs without an exception, but some answers are wrong.

Your job is to find the mistake using **printed program state**, then correct the function. Do not start by rewriting it.

1. Predict the result for each test by hand.
2. Run the test cell. Identify a passing case and a failing case.
3. Add a `print` **inside the loop** showing `value`, `current`, and `longest` after processing each day. Add a second `print` just before `return` if helpful.
4. Trace `[8, 8, 2, 7, 7, 7]` with `target = 7`. When does `current` grow? When does `longest` change?
5. Make the smallest correction, rerun the tests, and explain why it also works for `[]` and for a sequence entirely above the target.

````python
def longest_target_streak(values, target):
    current = 0
    longest = 0

    for value in values:
        if value >= target:
            current += 1
        else:
            if current > longest:
                longest = current
            current = 0

    return longest
````

````python
check(longest_target_streak([8, 8, 2], 7), 2)
check(longest_target_streak([8, 8, 2, 7, 7, 7], 7), 3)
check(longest_target_streak([7, 7], 7), 2)
check(longest_target_streak([2, 3, 1], 7), 0)
check(longest_target_streak([], 7), 0)
````

## 4.2. Debugging: increasing streak

The code below contains a logical error. `longest_increasing_run(values)` should return the number of values in the longest sequence of strictly increasing consecutive values.
For example, `[3, 5, 2, 4, 6]` has a longest run of length 3: `[2, 4, 6]`. An empty list has length 0. Equal values break a run.

````python
def longest_increasing_run(values):
    if not values:
        return 0

    previous = values[0]
    current = 1
    longest = 1

    for value in values[1:]:
        if value > previous:
            current += 1
            previous = value
        else:
            current = 1

        if current > longest:
            longest = current

    return longest
````

````python
check(longest_increasing_run([2, 4, 6, 1]), 3)
check(longest_increasing_run([3, 5, 2, 4, 6]), 3)
check(longest_increasing_run([9, 2, 3]), 2)
check(longest_increasing_run([5, 5, 3, 4, 6]), 3)
check(longest_increasing_run([4]), 1)
check(longest_increasing_run([]), 0)
````
