## A recipe for approaching a problem

Before writing any code, answer these questions:

1. **What are the inputs?** What are their types?
2. **What is the output?** What is its type?
3. **What assumptions can we make?** For example: words are separated by exactly one space.
4. **How would we solve one example by hand?**
5. **What information must we remember as we proceed?**
6. **Which edge cases should we test?**


````python
def check(actual, expected):
    """Display the result of a small test."""
    status = "✅" if actual == expected else "❌"
    print(f"{status} actual: {actual!r} | expected: {expected!r}")
````

### 3B. Manual version, without `split`

Now we cannot ask Python to separate the words automatically. We must detect where each word ends ourselves.


````python
def next_word_length(s, start):

  i = start
  length = len(s)
  count = 0

  while i < length and s[i] != ' ':
    count = count + 1
    i = i + 1

  return count



def filter_words_by_length_manual(s, min_length, max_length):
  result = ""
  length = len(s)
  i = 0
  while i < length:
    next_word_l =  next_word_length(s, i)
    if next_word_l >= min_length and next_word_l <= max_length:
      if len(result) > 0:
         result += ' '
      for k in range(next_word_l):
        result += s[i + k]
    i += (next_word_l+1)
  return result
````

````python
check(filter_words_by_length_manual("data drives digital change", 4, 6), "data drives change")
check(filter_words_by_length_manual("we learn python together", 5, 8), "learn python together")
check(filter_words_by_length_manual("a bb ccc", 4, 7), "")
check(filter_words_by_length_manual("", 1, 5), "")
check(filter_words_by_length_manual("one three seven", 5, 5), "three seven")
````
